# std in statistic = standard deviation

```python
cdf = Fa[['brand','model','year','fuel_type','horsepower','color','price_usd']] #گلچین کرد دیتایی که میخاد رو
cdf.head(7)
#فیچر هایی که من میخوام روش کار کنم  / دیتا ها رو کوچیک کردیم که مثلا روی همینا کار کنیم و توی یه متغیر ذخیرش کردیم
```

```python
viz = cdf[['horsepower','year']] #visualize یا به تصویر کشیدن
viz.hist() #histogram
```

```python
plt.scatter(cdf.horsepower, cdf.price_usd, color='red')
plt.xlabel('اسب بخار')
plt.ylabel('قیمت')
plt.show()
```

```python
msk = np.random.rand(len(Fa)) < 0.8 #میاد یه ماسک میسازه روی 
print(msk)
print(~msk) # not  اون رو نشون میده 
train =cdf[msk] #دیتای 80 درصد
test =cdf[~msk] #دیتای 20 درصد تست 
print(train)
print(test)
```

```python
fig = plt.figure() # ساخت یه فیگور جدید
ax1 = fig.add_subplot(111)
ax1.scatter(train.horsepower, train.price_usd, color='blue')
ax1.scatter(test.horsepower, test.price_usd, color='red')
plt.xlabel('اسب بخار')
plt.ylabel('قیمت دلار')
plt.show()
```
